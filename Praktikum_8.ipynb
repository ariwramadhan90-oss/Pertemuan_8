```python
scores = [80, 65, 90, 75, 70, 85]

# 1. Print the first and last score
print("First score:", scores[0])
print("Last score:", scores[-1])

# 2. Print the average and count scores above it
average = sum(scores) / len(scores)
print("Average:", average)

above_average = [score for score in scores if score > average]
print("Scores above average:", above_average)
print("Count above average:", len(above_average))

# 3. Print the position of the lowest score
lowest_score = min(scores)
position = scores.index(lowest_score) + 1
print("Lowest score:", lowest_score)
print("Position:", position)

# 4. Build a new list with every score + 5
new_scores = [score + 5 for score in scores]
print("New scores:", new_scores)
```
